# Retail Data Visualizations

This notebook reads `../data/Retail_Cleaned.csv` and creates the nine charts required by the APEX FDP Week 1 assignment.

Each chart is saved as a PNG in `../visualizations/` so the results can be viewed without running the notebook.

## 1. Imports and shared setup

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

PROJECT_DIR = Path.cwd()
if PROJECT_DIR.name.lower() == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent
elif not (PROJECT_DIR / "data").is_dir() and (PROJECT_DIR.parent / "data").is_dir():
    PROJECT_DIR = PROJECT_DIR.parent

DATA_FILE = PROJECT_DIR / "data" / "Retail_Cleaned.csv"
OUTPUT_DIR = PROJECT_DIR / "visualizations"
OUTPUT_DIR.mkdir(exist_ok=True)

df = pd.read_csv(DATA_FILE, parse_dates=["PurchaseDate"])
df["Month"] = df["PurchaseDate"].dt.to_period("M").astype(str)
sns.set_theme(style="whitegrid", palette="deep")

print(f"Loaded {len(df):,} cleaned rows")
print(f"Charts will be saved in: {OUTPUT_DIR}")
```

## 2. Customer demographics

### 2.1 Age distribution

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="Age", bins=15, kde=True, color="#176b87")
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Customers")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "01_age_distribution.png", dpi=150)
plt.show()
```

### 2.2 Gender distribution

```python
gender_counts = df["Gender"].value_counts()
plt.figure(figsize=(7, 5))
sns.barplot(x=gender_counts.index, y=gender_counts.values, color="#e07a5f")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Customers")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "02_gender_distribution.png", dpi=150)
plt.show()
```

### 2.3 Top 10 cities by customer count

```python
top_cities = df["City"].value_counts().head(10).sort_values()
plt.figure(figsize=(9, 5))
top_cities.plot(kind="barh", color="#3a86a8")
plt.title("Top 10 Cities by Customer Count")
plt.xlabel("Customers")
plt.ylabel("City")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "03_top_10_cities.png", dpi=150)
plt.show()
```

## 3. Sales insights

### 3.1 Total sales by product category

```python
category_sales = df.groupby("ProductCategory")["TotalAmount"].sum().sort_values()
plt.figure(figsize=(9, 5))
category_sales.plot(kind="barh", color="#2a9d8f")
plt.title("Total Sales by Product Category")
plt.xlabel("Sales amount")
plt.ylabel("Product category")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "04_sales_by_product_category.png", dpi=150)
plt.show()
```

### 3.2 Monthly sales trend

```python
monthly_sales = df.groupby("Month")["TotalAmount"].sum().sort_index()
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o", linewidth=2, color="#f4a261")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "05_monthly_sales_trend.png", dpi=150)
plt.show()
```

### 3.3 Payment mode usage

```python
payment_counts = df["PaymentMode"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(payment_counts.values, labels=payment_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Payment Mode Usage")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "06_payment_mode_usage.png", dpi=150)
plt.show()
```

## 4. Advanced insights

### 4.1 Average spend per customer by age group

```python
age_group_spend = df.groupby("AgeGroup", observed=False)["TotalAmount"].mean().dropna()
plt.figure(figsize=(8, 5))
sns.barplot(x=age_group_spend.index, y=age_group_spend.values, color="#8338ec")
plt.title("Average Spend by Age Group")
plt.xlabel("Age group")
plt.ylabel("Average transaction amount")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "07_average_spend_by_age_group.png", dpi=150)
plt.show()
```

### 4.2 City-wise revenue contribution

```python
city_revenue = df.groupby("City")["TotalAmount"].sum().sort_values()
plt.figure(figsize=(10, 6))
city_revenue.plot(kind="barh", color="#ff006e")
plt.title("City-wise Revenue Contribution")
plt.xlabel("Revenue")
plt.ylabel("City")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "08_city_wise_revenue.png", dpi=150)
plt.show()
```

### 4.3 Product category vs payment mode heatmap

```python
category_payment = pd.crosstab(df["ProductCategory"], df["PaymentMode"])
plt.figure(figsize=(9, 5))
sns.heatmap(category_payment, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Product Category vs Payment Mode")
plt.xlabel("Payment mode")
plt.ylabel("Product category")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "09_category_payment_heatmap.png", dpi=150)
plt.show()
```
